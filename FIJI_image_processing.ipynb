```python
# cell to reorganise all data into seperate folders within the image directory. 

import os
import shutil

# FUNCTIONS --------------------------------------------------------------------------------------------------
# create a new directory within image directory with a chosen name and move all files with ending with a phrase (e.g. "_metadata.csv") into that folder
def move_files_by_extension(input_folder, new_dir_name, ending):
    new_dir_path = os.path.join(input_folder, new_dir_name)
    os.makedirs(new_dir_path, exist_ok=True)
    for filename in os.listdir(input_folder):
        file_path = os.path.join(input_folder, filename)
        
        if os.path.isfile(file_path) and filename.endswith(ending):
            destination = os.path.join(new_dir_path, filename)
            shutil.move(file_path, destination)
            print(f"Moved: {filename}")

    print("Done.")

# MAIN SCRIPT -------------------------------------------------------------------------------------------------

# move metadata csvs
metadata_dir_name = "metadata"
move_files_by_extension(image_directory, metadata_dir_name, "_Merged.csv")

# move masks
mask_dir_name = "masks"
move_files_by_extension(image_directory, mask_dir_name, "_mask.tif")

# move mask info csvs
data_dir_name = "data"
move_files_by_extension(image_directory, data_dir_name, "_info.csv")
data_folder_path = os.path.join(image_directory, data_dir_name)
```

```python
# Cell to estimate the proportion of live cells remaining in each image and saves them to a "results" subfolder.
# This cell works with the nuclear and overlap (pixels in both the dead and nuclear masks) masks, and saves the results to a .csv file.

import os
import pandas as pd

# MAIN SCRIPT -------------------------------------------------------------------------------------------------

results_dir_name = "results"
results_dir_path = os.path.join(image_directory, results_dir_name)
os.makedirs(results_dir_path, exist_ok=True)

results = []
files = [f for f in os.listdir(data_folder_path) if f.endswith(".csv")]

nuclear_files = {}
dead_files = {}
for f in files:
    if f.endswith("_nuclear_mask_info.csv"):
        root = f.replace("_nuclear_mask_info.csv", "")
        nuclear_files[root] = f
    elif f.endswith("_overlap_mask_info.csv"):
        root = f.replace("_overlap_mask_info.csv", "")
        dead_files[root] = f
common_roots = set(nuclear_files.keys()) & set(dead_files.keys())

for root in common_roots:
    nuclear_path = os.path.join(data_folder_path, nuclear_files[root])
    dead_path = os.path.join(data_folder_path, dead_files[root])

    nuclear_df = pd.read_csv(nuclear_path)
    dead_df = pd.read_csv(dead_path)

    total_nuclear_area = nuclear_df["Area"].sum()
    total_dead_area = dead_df["Area"].sum()
    perc_live_remaining = (1 - (total_dead_area/total_live_area))*100

    results.append({
        "root_name": root,
        "total_nuclear_area": total_nuclear_area,
        "total_dead_area": total_dead_area,
        "percentage_live_cells_remaining": perc_live_remaining
    })

result_df = pd.DataFrame(results)
result_csv_path = os.path.join(results_dir_path, "results.csv")
result_df.to_csv(result_csv_path, index=False)

print(result_df)
```
