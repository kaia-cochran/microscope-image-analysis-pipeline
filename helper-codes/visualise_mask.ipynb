```python
# code to visualise the masks produced by the pipeline for QCing.
# takes mask files and overlays them on the relevant .tif image so you can see what is being measured. 
# this cell does mask overlays for entire dataset.

import os
import matplotlib.pyplot as plt
from skimage import io
from skimage.segmentation import find_boundaries
import torch

# INPUTS ------------------------------------------------------------------------------------------------------

# image directory containing all images and masks (the same directory you inputted in pipeline)
image_directory = r"Path\to\image_directory

# FUNCTIONS ----------------------------------------------------------------------------------------------------------
def dead_mask_overlay(output_folder, image_path, dead_mask_path, image_name):
    image = io.imread(image_path)
    dead_mask = io.imread(dead_mask_path)
    dead_outline = find_boundaries(dead_mask, mode="outer")
    
    output_name = image_name +  "_dead_overlay.tif"
    output = os.path.join(output_folder, output_name)

    fig, ax = plt.subplots(figsize=(6,6))
    ax.imshow(image, cmap="gray")
    ax.contour(dead_outline, colors="red", linewidths=0.1)
    ax.axis("off")

    plt.savefig(output, bbox_inches="tight", pad_inches=0, dpi=300)
    plt.close()

    print("saved dead overlay: ", output)

def nuclear_mask_overlay(output_folder, image_path, nuclear_mask_path, image_name):
    image = io.imread(image_path)
    nuclear_mask = io.imread(nuclear_mask_path)
    nuclear_outline = find_boundaries(nuclear_mask, mode="outer")
    
    output_name = image_name +  "_nuclear_overlay.tif"
    output = os.path.join(output_folder, output_name)

    fig, ax = plt.subplots(figsize=(6,6))
    ax.imshow(image, cmap="gray")
    ax.contour(nuclear_outline, colors="cyan", linewidths=0.1)
    ax.axis("off")

    plt.savefig(output, bbox_inches="tight", pad_inches=0, dpi=300)
    plt.close()

    print("saved nuclear overlay: ", output)

def overlap_mask_overlay(output_folder, image_path, overlap_mask_path, image_name):
    image = io.imread(image_path)
    overlap_mask = io.imread(overlap_mask_path)
    overlap_outline = find_boundaries(overlap_mask, mode="outer")
    
    output_name = image_name +  "_overlap_overlay.tif"
    output = os.path.join(output_folder, output_name)

    fig, ax = plt.subplots(figsize=(6,6))
    ax.imshow(image, cmap="gray")
    ax.contour(overlap_outline, colors="green", linewidths=0.1)
    ax.axis("off")

    plt.savefig(output, bbox_inches="tight", pad_inches=0, dpi=300)
    plt.close()

    print("saved overlap overlay: ", output)

# function to connect notebook to GPU
def connect_gpu():
    if torch.cuda.is_available():
        device = torch.device("cuda")
    else:
        device = torch.device("cpu")
    print("Using:", device)

# MAIN SCRIPT --------------------------------------------------------------------------------------------------------
connect_gpu()

image_ending = "_Merged.tif"
dead_mask_ending = "_apoptosis_mask.tif"
nuclear_mask_ending = "_nuclear_mask.tif"
overlap_mask_ending = "_overlap_mask.tif"

output_dir_name = "mask_overlays"
output_dir_path = os.path.join(image_directory, output_dir_name)
os.makedirs(output_dir_path, exist_ok=True)

images = {}
nuclear_masks = {}
dead_masks = {}
overlap_masks = {}

for root_dir, subdirs, files in os.walk(image_directory):
    for f in files:

        if f.endswith(image_ending):
            root = os.path.splitext(f)[0]
            full_path = os.path.join(root_dir, f)
            images[root] = full_path

        elif f.endswith(dead_mask_ending):
            root = f.replace(dead_mask_ending, "")
            full_path = os.path.join(root_dir, f)
            dead_masks[root] = full_path

        elif f.endswith(nuclear_mask_ending):
            root = f.replace(nuclear_mask_ending, "")
            full_path = os.path.join(root_dir, f)
            nuclear_masks[root] = full_path

        elif f.endswith(overlap_mask_ending):
            root = f.replace(overlap_mask_ending, "")
            full_path = os.path.join(root_dir, f)
            overlap_masks[root] = full_path

common_roots = set(images.keys()) & set(dead_masks.keys()) & set(nuclear_masks.keys()) & set(overlap_masks.keys())

for root in common_roots:

    image_path = images[root]
    dead_mask_path = dead_masks[root]
    nuclear_mask_path = nuclear_masks[root]
    overlap_mask_path = overlap_masks[root]
    dead_mask_overlay(output_dir_path, image_path, dead_mask_path, root)
    nuclear_mask_overlay(output_dir_path, image_path, nuclear_mask_path, root)
    overlap_mask_overlay(output_dir_path, image_path, overlap_mask_path, root)

print("done")
```
