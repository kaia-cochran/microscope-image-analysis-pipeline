```python
# Code to convert spaces in all subfolder/file names in a directory to underscores for python reading. No data will be altered, just the names.
# Run this before running the pipeline on any new datasets

import os

# INPUTS ----------------------------------------------------------------------------------------------------
# directory whose paths/files will be renamed
image_directory = r"\path\to\image_directory"

# MAIN SCRIPT ----------------------------------------------------------------------------------------------------
for root, dirs, files in os.walk(image_directory, topdown=False):
    
    # Rename folders
    for name in dirs:
        if " " in name:
            old_path = os.path.join(root, name)
            new_name = name.replace(" ", "_")
            new_path = os.path.join(root, new_name)

            if os.path.exists(new_path):
                print(f"Skipped: '{old_path}' to '{new_path}' (target path already exists)")
            else:
                os.rename(old_path, new_path)
                print(f"Renamed: '{old_path}' to '{new_path}'")

    # Rename files
    for name in files:
        if " " in name:
            old_path = os.path.join(root, name)
            new_name = name.replace(" ", "_")
            new_path = os.path.join(root, new_name)

            if os.path.exists(new_path):
                print(f"Skipped: '{old_path}' to '{new_path}' (target path already exists)")
            else:
                os.rename(old_path, new_path)
                print(f"Renamed file: '{old_path}' to '{new_path}'")

print("done")
```
