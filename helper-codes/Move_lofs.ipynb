```python
# This code moves all relevant lof files within image subdirectories out to the main image directory so they can be run through the pipeline.
# The files are renamed to a concatenation of their subdirectory names and original image name to remain identifiable. For example:
# an image named "image_merged.lof" found through the path "image_directory" -> "date_time_assay" -> "sequence001" -> "A" wil be renamed to:
# "date_time_assay_sequence001_A_image_merged.lof" and found in "image_directory".
# Run this before running the pipeline on any new datasets

import os
import shutil

# INPUTS -------------------------------------------------------------------------

# directory containing your .lof files
image_directory = r"\Path\to\image_directory"

# the file ending you want to search for, can just be ".lof":
file_ending = "_merged.lof"

# FUNCTIONS ---------------------------------------------------------------------

def flatten_lof_files(root_dir, ending):
    root_dir = os.path.abspath(root_dir)

    for current_dir, dirs, files in os.walk(root_dir):
        if current_dir == root_dir:
            continue

        rel_path = os.path.relpath(current_dir, root_dir)
        folders = rel_path.split(os.sep)

        for file in files:
            if file.lower().endswith(ending"):
                old_path = os.path.join(current_dir, file)

                name, ext = os.path.splitext(file)
                new_name = "_".join(folders + [name]) + ext
                new_path = os.path.join(root_dir, new_name)

                # Skip if destination file already exists - safeguard in order to not accidentally overwrite existing files
                if os.path.exists(new_path):
                    print(f"Skipping (already exists): {new_name}")
                    continue

                shutil.move(old_path, new_path)
                print(f"Moved: {old_path} to {new_path}")

# MAIN SCRIPT --------------------------------------------------------------------------
flatten_lof_files(image_directory)
```
